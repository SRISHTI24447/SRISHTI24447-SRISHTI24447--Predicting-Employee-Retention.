### **7.3 Find the optimal cutoff** <font color = red>[15 marks]</font>

Find the optimal cutoff to improve model performance. While a default threshold of 0.5 was used for initial evaluation, optimising this threshold can enhance the model's performance.

First, plot the ROC curve and check AUC.



7.3.1 Plot ROC curve <font color="red">[3 Marks]</font>

```python
# Define ROC function
def plot_roc_curve(fpr, tpr):
    plt.plot(fpr, tpr, color='orange', label='ROC')
    plt.plot([0, 1], [0, 1], color='darkblue', linestyle='--')
    plt.xlabel('False Positive Rate')
    plt.ylabel('True Positive Rate')
    plt.title('Receiver Operating Characteristic (ROC) Curve')
    plt.legend()
    plt.show()
```

```python
# Call the ROC function
from sklearn.metrics import roc_curve, auc
fpr, tpr, thresholds = roc_curve(df_train['Actual'], df_train['Predicted'])
roc_auc = auc(fpr, tpr)
plot_roc_curve(fpr, tpr)
```

**Sensitivity and Specificity tradeoff**

Check sensitivity and specificity tradeoff to find the optimal cutoff point.

7.3.2 Predict on training set at various probability cutoffs <font color="red">[1 Mark]</font>

```python
# Predict on training data by creating columns with different probability cutoffs to explore the impact of cutoff on model performance
from sklearn.metrics import roc_curve
df_train['Predicted_prob'] = y_train_pred
df_train.head()
```

7.3.3 Plot for accuracy, sensitivity, specificity at different probability cutoffs <font color="red">[2 Marks]</font>

```python
# Create a DataFrame to see the values of accuracy, sensitivity, and specificity at different values of probability cutoffs
cutoff_values = []
accuracy = []
sensitivity = []
specificity = []

```

```python
# Plot accuracy, sensitivity, and specificity at different values of probability cutoffs
for i in thresholds:
    # Use 'Predicted_prob' column from df_train for predictions
    y_val_pred_new = [1 if prob > i else 0 for prob in df_train['Predicted_prob']]
    cm = metrics.confusion_matrix(df_train['Actual'], y_val_pred_new)
    tn, fp, fn, tp = cm.ravel()

    acc = (tp + tn) / (tp + tn + fp + fn)
    sens = tp / (tp + fn) if (tp + fn) > 0 else 0
    spec = tn / (tn + fp) if (tn + fp) > 0 else 0

    cutoff_values.append(i)
    accuracy.append(acc)
    sensitivity.append(sens)
    specificity.append(spec)

# Create DataFrame
cutoff_df = pd.DataFrame({
    'Cutoff': cutoff_values,
    'Accuracy': accuracy,
    'Sensitivity': sensitivity,
    'Specificity': specificity
})

# Plotting
plt.figure(figsize=(10, 6))
plt.plot(cutoff_df['Cutoff'], cutoff_df['Accuracy'], label='Accuracy', color='blue')
plt.plot(cutoff_df['Cutoff'], cutoff_df['Sensitivity'], label='Sensitivity', color='green')
plt.plot(cutoff_df['Cutoff'], cutoff_df['Specificity'], label='Specificity', color='red')
plt.xlabel('Probability Cutoff')
plt.ylabel('Score')
plt.title('Accuracy, Sensitivity, Specificity at Different Cutoffs')
plt.legend()
plt.grid(True)
plt.show()
```
